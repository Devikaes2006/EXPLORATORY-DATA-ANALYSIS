```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
df=pd.read_csv("/titanic.csv")
df
df.head()
print(df.shape)# to show no of rows and columns
df.info()
df.describe()#give statistical summary of numerical operation(mean,mode etc)
df.isnull().sum()
df["Embarked"]=df["Embarked"].fillna(  # REPLACE MISSING VALUES
    df["Embarked"].mode()[0]   #FIND THE MOST FREQUENT VALUE
)
print(df.isnull().sum())
print("before:",df.shape[0])
df=df.drop_duplicates()
print("after:",df.shape[0])
df=df.rename(columns={
    "passengerid":"passengerID",
    "2urvived":"survived"
})
df["Sex"]=df["Sex"].map({
    0:"Male",
    1:"Female"
})
df.head()
sns.histplot(data=df,x="Age",bins=20)
plt.title("age distribution of titanic passengers")
plt.xlabel("Age")
plt.ylabel("number of passengers")
plt.show()
sns.countplot(data=df,x="Sex",hue="survived")
plt.title("titanic survival by sex")
plt.xlabel("sex")
plt.ylabel("number of passengers")
plt.show()
sns.countplot(data=df,x="Pclass",hue="survived")
plt.title("titanic survival by passenger class")
plt.xlabel("passenger class")
plt.ylabel("number of passenger")
plt.show()
```
